```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
x0 = 0.0  # Initial x position
y0 = 1000.0  # Initial y position
v0x = 20.0  # Initial x velocity (m/s)
v0y = 0.0  # Initial y velocity (m/s)
g = 9.81  # Acceleration due to gravity (m/s^2)

# Time values
t = np.linspace(0, y0, num=100)  # Adjust the time range as needed

# Calculate x and y positions
x = x0 + v0x * t
y = y0 + v0y * t - 0.5 * g * t**2

# Plot the motion
plt.figure(figsize=(8, 6))
plt.plot(x, y)
plt.xlabel("Horizontal Distance (m)")
plt.ylabel("Vertical Distance (m)")
plt.title("Projectile Motion of a Falling Object")
plt.grid(True)
plt.show()

```
